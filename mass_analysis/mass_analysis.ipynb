# $m_\pi$ and $m_{\mathrm{PCAC}}$ analysis

Program for measuring the pion and PCAC mass given the correlators  $$c(n_t):=\frac{1}{\sqrt{N_x}}\sum_x \langle \mathcal{O}_\pi(x,t) \overline{\mathcal{O}}_\pi(0,0)\rangle=-\sum_x\sum_{\alpha,\beta=0}^1 |D^{-1}(x,t|0,0)_{\alpha,\beta}|^2.$$

$$m_{\mathrm{PCAC}}(t) = \frac{C_{\mathrm{AP}}(t+1)-C_{\mathrm{AP}}(t-1)}{4 C_{\mathrm{PP}}(t)}$$
$$C_{\mathrm{PP}}(t) = -\frac{1}{2 \sqrt{N_x}} \sum_x \sum_{\alpha,\beta=0}^1 |D^{-1}(x,t|0,0)_{\alpha,\beta}|^2$$
$$C_{\mathrm{AP}}(t) = -\frac{1}{2 \sqrt{N_x}} \sum_x \mathrm{Tr}\left[\gamma_0 \gamma_5 D^{-1}(x,t|0,0) \gamma_5 D^{-1}(0,0|x,t)\right]$$
